```python
# 1) WAP to reverse the given string

s="python"

reverse=(s[::-1])
print("reverse=",reverse)
```

```python
# 2) check whether a string is palindrome

s="madam"

reverse=(s[::-1])
if s == reverse:
    print("palindrome")
else:
    print("not palindrome")
```

```python
# 3) WAP to count consonants in a given string

s="python"

vowels="aeiou"
count=0

for letter in s:
    if letter not in  vowels:
        count = count + 1
print("number of consonants=",count)
```

```python
# 4) WAP to count spaces in a given string

s="python programming language"

count=0
for letter in s :
    if letter==" ":
     count += 1
print("number of spaces=",count)
```

```python
# 5) WAP to count how many times "a" occurs in a string

s="banana"
count=0
for letter in s:
    if letter=="a":
     count = count + 1
print("number of a=",count)
```

```python
# 6) WAP to find the length of a given string

s="pradnya rasal"

print(len(s))
```

```python
# 7) WAP to check whether 'a' is present in the given string.

s="instagram"
if "a" in s:
    print("a is present")
else:
    print("a is not present")
```

```python
# 8) WAP to count digits in a given string.

s="python1234"

count=0

for letter in s:
    if letter.isdigit():
        count=count+1
print("number of digits=",count)
```

```python
# 9) WAP to convert a given string into uppercase.

s="pradnya rasal"
print(s.upper())
```

```python
# 10) count vowels and consonants

s='developer'

vowels=0
consonants=0
for letter in s:
    if letter in "aeiou":
        vowels=vowels+1
    else:
        consonants=consonants+1
print("vowels=",vowels)
print("consonants=",consonants)
```

```python
# 11) find "m" in string

s="programming"

for letter in s:
    if letter=="m":
      print(letter,"found")
    else:
      print(letter,"not found m")
```

```python
# 12) count even and odd numbers

numbers=[10,15,22,31,40,47]

even=0
odd=0

for number in numbers:
    if number % 2==0:
        even=even+1
        print("even")

    else:
        odd=odd+1
        print("odd")
```
